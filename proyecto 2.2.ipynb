# Ejercicio 1: Orden de los números
Lee tres números por teclado. Evalúa si los números están en orden ascendente o no. Si el primer número introducido es 0, muestra un mensaje de error.

```python
# Orden de tres números sin usar listas.
# Entrada de datos: Tres números enteros. 
num1 = int(input("Ingrese el primer número: "))
num2 = int(input("Ingrese el segundo número: "))
num3 = int(input("Ingrese el tercer número: "))
# Validación del numero 1, comprobar si es 0. 
if num1 == 0:
    print("El primer número no puede ser cero.")
else: 
    # Comprobar el orden ascendente de los números. 
    if num1 < num2 and num2 < num3: 
        print(f"Los números {num1}, {num2} y {num3} están en orden ascendente.")
    else:
        print(f"Los números {num1}, {num2} y {num3} no están en orden ascendente.")
```

# Ejercicio 3: Sumatorio de los primeros N numeros. 
Calculamos la suma acumulada de los números en el rango de [0,N−1] utilizando un bucle for y la función range(N)

```python
# Sumatorio de los primeros N números (desde 0 hasta N-1)
N = int(input("Ingrese un número entero positivo N:"))
# Validación del número N, comprobar si es entero y positivo. 
if N < 0:
    print("El número N debe ser un entero positivo, por favor, ingrese un número válido.")
else: 
    # Iniciamos la variable sumatoria en 0.
    sumatoria = 0
    terminos = []
    # Bucle para sumar los números desde el cero hasta N-1.
    for i in range(N):
        sumatoria += i
        terminos.append(str(i))
    # Unimos los términos con + 
    expresion = " + ".join(terminos)
    print(f"La sumatoria de los primeros {N} números es: {expresion} = {sumatoria}")
    print(f"La sumatoria de los primeros {N} números es: {sumatoria}")
    print(f"Con N = {N} → sumatorio = {sumatoria}")
```

# Ejercicio 4: Contador de letras "a"
Implementamos un bucle while True que solicita caracteres individuales. Contamos las apariciones de la letra 'a' (y 'A') y finalizamos la lectura al detectar un punto (.)

```python
# Contador de letras "a" caracter por caracter en una cadena de texto. 
contador_a = 0 
fin_de_lectura = False
while not fin_de_lectura:
    cadena = input("Ingrese una cadena de texto (o finalice con un '.'): ")
    # Imprimimos la frase ingresada por el usuario. 
    print(f"Cadena ingresada: {cadena}")
# Recorremos cada caracter de la cadena ingresada
    for caracter in cadena:
       if caracter == ".": 
             fin_de_lectura = True
             break
        # Comprobamos si es 'a' o 'A' usando lower()
       if caracter.lower() == "a":
            contador_a += 1
print(f"El número total de letras 'a' en las cadenas ingresadas es: {contador_a}")
```

# Ejercicio 6: Sumatorio de números pares decrecientes
Generamos una secuencia decreciente de números pares entre 1000 y 73 utilizando range dentro de un bucle for y acumulamos su suma.

```python
# Numeros pares decrecientes entre 1000 y 73
suma_pares = 0 
# usamos range para generar los numeros pares en orden decreciente. 
for numero in range(1000,72,-1):
    if numero % 2 == 0:
        suma_pares += numero
print(f"La suma de los números pares entre 1000 y 73 es: {suma_pares}")
```

# Ejercicio 7: Lista de números: análisis y operaciones
Leemos una cantidad variable de números, los guardamos en una lista y aplicamos métodos nativos como .sort(), sum(), max(), min(), junto con .index() para localizar las posiciones.

```python
# Análisis estadístico básico de una lista de números.
# Creamos una lista vacía para ala¡macenar los números ingresados por el usuario. 
numeros_lista = []
# Preguntamos al usuario cuántos números desea ingresar. 
cantidad_numeros = int(input ("Ingrese la cantidad de números que desea analizar: "))
# Validación de la cantidad de números, debe ser un entero positivo. 
if cantidad_numeros <= 0:
    print("La cantidad de números debe ser un entero positivo, por favor, ingrese un número válido.")
else:
    # Bucle para ingresar los números a la lista. 
    for i in range(cantidad_numeros):
        numero = int(input(f"Ingrese el número {i+1}:"))
        numeros_lista.append(numero)
    # Imprimimos la lista de números ingresada por el usuario. 
    print("Lista de números ingresada:" , numeros_lista)
    # 1. máximo y mínimo de la lista.
    maximo = max(numeros_lista)  
    minimo = min(numeros_lista)  
    # 2. Identifica la posición del número máximo y mínimo en la lista. 
    posicion_maximo = numeros_lista.index(maximo)
    posicion_minimo = numeros_lista.index(minimo)
    # 3. Ordenamos la lista de menor a mayor. 
    numeros_lista.sort()
    # 4. Calculamos el promedio de los números antes de ordenarlos. 
    promedio = sum(numeros_lista) / len(numeros_lista)
    # Imprimimos los resultados del análisis estadístico. 
    print(f"Lista de números ordenada: {numeros_lista}")
    print(f"Promedio de los números: {promedio}")
    print(f"Máximo de los números: {maximo}")
    print(f"Mínimo de los números: {minimo}")
    print(f"Posición de máximo: {posicion_maximo + 1}") # Sumamos 1 para mostrar la posición en lenguaje natural
    print(f"Posición de mínimo: {posicion_minimo + 1}") # Sumamos 1 para mostrar la posición en lenguaje natural
```

# Ejercicio 8: Menú compra-venta
Sistema interactivo de gestión de transacciones

```python
# Creamos el menú con la estructura while para que el usuario pueda elegir entre las diferentes opciones. 
# Creamos una lista vacía con alquileres, ventas y compras. 
alquileres = []
ventas = []
compras = []
# Con el bucle while podremos preguntar al usuario y almacenar su respuesta en la lista correspondiente. 
while True: 
    print("\n" + "="*20)
    print(" MENÚ COMPRA-VENTA")
    print("="*20)
    print("1. Comprar")
    print("2. Vender")
    print("3.Alquilar")
    print("4. Mostrar listas de datos")
    print("5. Salir")
    opcion = input("Seleccione una opción (1-5): ")
    if opcion == "1": 
        articulo = input("¿Qué desea comprar? ")
        compras.append(articulo)
        print(f"Artículo '{articulo}' agregado a la lista de compras.")
    elif opcion == "2": 
        articulo = input("¿Qué desea vender? ")
        ventas.append(articulo)
        print(f"Artículo '{articulo}' agregado a la lista de ventas.")
    elif opcion == "3":
        articulo = input("¿Qué desea alquilar? ")
        alquileres.append(articulo)
        print(f"Artículo '{articulo}' agregado a la lista de alquileres.")
    elif opcion == "4":
        print("\n ---Resumen de listas:---")
        # Verificamos si hay datos guardados para evitar imprimir listas vacías.
        if not compras:
            print("Compras: [vacía]")
        else:
            for i, elem in enumerate(compras, start=1):
                print(f"Compras: {i}. {elem}")
        if not ventas:
            print("Ventas: [vacía]")
        else:
            for i, elem in enumerate(ventas, start=1):
                print(f"Ventas: {i}. {elem}")
        if not alquileres:
            print("Alquileres: [vacía]")
        else:
            for i, elem in enumerate(alquileres, start=1):
                print(f"Alquileres: {i}. {elem}")
    elif opcion == "5": 
        print("Gracias por usar el programa. ¡Hasta pronto!")
        break
    else: 
        print("Opción no válida. Por favor, seleccione una opción del 1 al 5.")
```

# Ejercicio 9: Programa de login con reintentos
Implementamos un sistema de autenticación con doble bucle: el bucle externo permite reiniciar la configuración del sistema si se agotan los 5 intentos de inicio de sesión de forma consecutiva.

```python
# Control de acceso con reintentos y reinicio de sesión.
print("\n--- CONFIGURACIÓN DEL SISTEMA ---")

# 1. Configuración inicial (se hace una sola vez)
password_correcta = input("Establece la contraseña para el sistema: ")
print(f"Contraseña establecida: {password_correcta}")

# 2. Bucle principal que permite reiniciar la sesión
while True:
    intentos = 0 
    Login_exitoso = False

    # Bucle de reintentos
    while intentos < 5: 
        intentos += 1
        password_ingresada = input("Ingrese la contraseña para acceder al sistema: ")
        
        if password_ingresada == password_correcta:
            print("Login Correcto.")
            Login_exitoso = True
            break  # Sale del bucle de intentos al acertar
        else: 
            print(f"Login Incorrecto. Intento {intentos}/5.")

    # 3. Evaluación de la sesión
    if Login_exitoso:
        print("Acceso concedido. Bienvenido al sistema.")
        break  # Termina el programa definitivamente
    else:
        print("\nSe han agotado los intentos. Reiniciando sesión...\n")
        # Al no haber break aquí, el bucle 'while True' vuelve a empezar
```

# Ejercicio 10: Verificar aprobados y suspendidos con un diccionario.
Recorremos un diccionario de estudiantes mediante el método .items() para extraer la clave (nombre) y el valor (calificación), aplicando un condicional para evaluar si aprueban o suspenden.

```python
# Evaluación de calificaciones de estudiantes con bucle while y control de flujo desde un diccionario. 
notas = { "Ana": 5, "Luis": 3, "Marta": 8, "Carlos": 7, "Lucía": 4, "Javier": 6, "Sofía": 9}
print("Calificaciones Lengua y Literatura:")
for estudiante, nota in notas.items():
    if nota >= 5: 
       estado = "Aprobado"
    else:
       estado = "Suspendido"
    print(f"{estudiante}: {nota} - {estado}")
```
