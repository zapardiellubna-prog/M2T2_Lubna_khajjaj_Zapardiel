# Ejercicio 1: Orden de los números
Lee tres números por teclado. Evalúa si los números están en orden ascendente o no. Si el primer número introducido es 0, muestra un mensaje de error.

```python
# Orden de tres números sin usar listas.
# Entrada de datos: Tres números enteros. 
num1 = int(input("Ingrese el primer número: "))
num2 = int(input("Ingrese el segundo número: "))
num3 = int(input("Ingrese el tercer número: "))
# Validación del numero 1, comprobar si es 0. 
if num1 == 0:
    print("El primer número no puede ser cero.")
else: 
    # Comprobar el orden ascendente de los números. 
    if num1 < num2 and num2 < num3: 
        print(f"Los números {num1}, {num2} y {num3} están en orden ascendente.")
    else:
        print(f"Los números {num1}, {num2} y {num3} no están en orden ascendente.")
```

# Ejercicio 3: Sumatorio de los primeros N numeros. 
Calculamos la suma acumulada de los números en el rango de [0,N−1] utilizando un bucle for y la función range(N)

```python
# Sumatorio de los primeros N números (desde 0 hasta N-1)
N = int(input("Ingrese un número entero positivo N:"))
# Validación del número N, comprobar si es entero y positivo. 
if N < 0:
    print("El número N debe ser un entero positivo, por favor, ingrese un número válido.")
else: 
    # Iniciamos la variable sumatoria en 0.
    sumatoria = 0
    terminos = []
    # Bucle para sumar los números desde el cero hasta N-1.
    for i in range(N):
        sumatoria += i
        terminos.append(str(i))
    # Unimos los términos con + 
    expresion = " + ".join(terminos)
    print(f"La sumatoria de los primeros {N} números es: {expresion} = {sumatoria}")
    print(f"La sumatoria de los primeros {N} números es: {sumatoria}")
    print(f"Con N = {N} → sumatorio = {sumatoria}")
```

# Ejercicio 4: Contador de letras "a"
Implementamos un bucle while True que solicita caracteres individuales. Contamos las apariciones de la letra 'a' (y 'A') y finalizamos la lectura al detectar un punto (.)

```python
# Contador de letras "a" caracter por caracter en una cadena de texto. 
contador_a = 0 
fin_de_lectura = False
while not fin_de_lectura:
    cadena = input("Ingrese una cadena de texto (o finalice con un '.'): ")
    # Imprimimos la frase ingresada por el usuario. 
    print(f"Cadena ingresada: {cadena}")
# Recorremos cada caracter de la cadena ingresada
    for caracter in cadena:
       if caracter == ".": 
             fin_de_lectura = True
             break
        # Comprobamos si es 'a' o 'A' usando lower()
       if caracter.lower() == "a":
            contador_a += 1
print(f"El número total de letras 'a' en las cadenas ingresadas es: {contador_a}")
```

# Ejercicio 5: Contador de letras "a" en palabras completas.

# Ejercicio 6: Sumatorio de números pares decrecientes
Generamos una secuencia decreciente de números pares entre 1000 y 73 utilizando range(1000, 72, -2) dentro de un bucle for y acumulamos su suma.

```python
# Numeros pares decrecientes entre 1000 y 73
suma_pares = 0 
# usamos range para generar los numeros pares en orden decreciente. 
for numero in range(1000,72,-1):
    if numero % 2 == 0:
        suma_pares += numero
print(f"La suma de los números pares entre 1000 y 73 es: {suma_pares}")
```
